```python
df["Check-in Date"] = pd.to_datetime(df["Check-in Date"])
df["Check-out Date"] = pd.to_datetime(df["Check-out Date"])

df.dtypes
```

```python
print("Number of Rooms:", df["Room Number"].nunique())
print("Minimum Room Number:", df["Room Number"].min())
print("Maximum Room Number:", df["Room Number"].max())
```

```python
df["Check-in Date"] = pd.to_datetime(df["Check-in Date"])
df["Check-out Date"] = pd.to_datetime(df["Check-out Date"])

start_date = df["Check-in Date"].min()
end_date = df["Check-out Date"].max()

# The dataset contains 100 unique rooms. Calculate this from the data
# instead of hard-coding the value.
total_rooms = df["Room Number"].nunique()
print("Total Rooms:", total_rooms)

dates = pd.date_range(start=start_date, end=end_date - pd.Timedelta(days=1))

daily_occupancy = []

for date in dates:
    occupied_rooms = df[
        (df["Check-in Date"] <= date) &
        (df["Check-out Date"] > date)
    ]["Room Number"].nunique()

    occupancy_rate = (occupied_rooms / total_rooms) * 100

    daily_occupancy.append({
        "Date": date,
        "Occupied Rooms": occupied_rooms,
        "Occupancy Rate": occupancy_rate
    })

occupancy_df = pd.DataFrame(daily_occupancy)

occupancy_df.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    occupancy_df["Date"],
    occupancy_df["Occupancy Rate"]
)

plt.xlabel("Date")
plt.ylabel("Occupancy Rate (%)")
plt.title("Daily Hotel Occupancy Rate")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
occupancy_df["Day"] = occupancy_df["Date"].dt.day
occupancy_df["Month"] = occupancy_df["Date"].dt.month
occupancy_df["Year"] = occupancy_df["Date"].dt.year
occupancy_df["DayOfWeek"] = occupancy_df["Date"].dt.dayofweek
```

```python
occupancy_df.head()
```

### Important modeling correction
`Occupied Rooms` is not used as an input feature because it directly determines `Occupancy Rate`. Using it would cause target leakage. The model therefore uses historical occupancy (`Lag_1` and `Lag_7`) together with calendar features. The train/test split is chronological because the target is time-dependent.

```python
occupancy_df["Lag_1"] = occupancy_df["Occupancy Rate"].shift(1)
occupancy_df["Lag_7"] = occupancy_df["Occupancy Rate"].shift(7)

occupancy_df = occupancy_df.dropna().copy()

X = occupancy_df[["Day", "Month", "Year", "DayOfWeek", "Lag_1", "Lag_7"]]
y = occupancy_df["Occupancy Rate"]
```

```python
X.head()
```

```python
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()
y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeRegressor

dt_model = DecisionTreeRegressor(
    random_state=42,
    max_depth=5
)

dt_model.fit(X_train, y_train)
```

```python
y_pred = dt_model.predict(X_test)

print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    max_depth=10
)

rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)

print(y_pred_rf[:10])
```

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("MAE :", mae_rf)
print("MSE :", mse_rf)
print("RMSE:", rmse_rf)
print("R²  :", r2_rf)
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
}).sort_values(
    by="Importance",
    ascending=False
)

feature_importance
```
